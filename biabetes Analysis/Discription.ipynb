## Diabetes Analytics Dashboard – Power BI 🩺📊
* This practice dashboard is built for Data Analytics, Data Visualization, and Data Science learning. It provides meaningful insights into diabetes risk factors using interactive visuals and advanced analytics.

1) 🔹 Key Metrics – Total patients, BMI, glucose, blood pressure, and insulin levels.
2) 🔹 Diabetes Risk Segmentation – Categorized into High, Medium, and Low risk groups.
3) 🔹 Trends & Distribution – Glucose vs. Age, BMI categories, and Blood Pressure analysis.
4) 🔹 Correlation Analysis – Exploring the relationships between glucose, BMI, and diabetes risk.
5) 🔹 Gauge & Pie Charts – Visualizing risk percentage, BMI distribution, and glucose levels.
6) 🔹 Interactive Filters & Drilldowns – Allowing deeper exploration of specific patient groups.
7) 🔹 Predictive Insights – Identifying potential risk patterns through visual analytics.

* This project helps in understanding data-driven healthcare insights using Power BI.
